## 1. What the AI would see about the SomaScan file

The sample columns (if their names differ only in digits) are one *template* entry. Text columns with mostly distinct values (SeqId, Target) send only their shape; columns with few values (Organism, Type) send values with counts.

```python
soma = prism.read(f"{DATA}/SomaExpr_common.csv")
d = prism.digest(soma, prism.propose_layout(soma))
prism.show(d)
```

## 2. The metabolomics file

1,174 metabolite columns named like `Amino Acid_35` become one entry per class.

```python
meta = prism.read(f"{DATA}/MetaboExpr_common.csv")
d2 = prism.digest(meta, prism.propose_layout(meta))
prism.show(d2)
```

## 3. Without any example values

If you prefer that no cell value at all leaves Colab, use `examples=False`: only names and statistics are sent.

```python
prism.show(prism.digest(soma, examples=False), max_chars=1500)
```
