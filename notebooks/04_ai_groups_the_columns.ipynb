# PRISM step 4: the AI groups the columns

The AI reads the digest (step 3: names and statistics, no data rows) and proposes which columns belong together and what each group is. Code then checks the answer: every column named must exist and is claimed once, columns the AI left out become *unresolved* groups, and anything refused is listed with its reason. Nothing is confirmed yet: that is the next step.
